```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Make a copy of the DataFrame
df_impute = df.copy()

# Categorical columns with missing values
missing_categoricals = ['income', 'race', 'dnr', 'sfdm2']

# Loop through each categorical column that has missing values
for col in missing_categoricals:
    # Create mask for missing values
    missing_mask = df[col].isnull()

    # Define features for training (exclude target column and non-numeric columns)
    feature_cols = df.columns.difference(missing_categoricals + ['id'])

    # Label encode the target column (ensuring all possible values are included)
    le = LabelEncoder()
    df_impute[col] = le.fit_transform(df[col].astype(str).fillna('missing'))

    # Encode the feature columns too (non-numeric features)
    df_impute_encoded = df_impute[feature_cols].copy()
    for feature in feature_cols:
        if df_impute[feature].dtype == 'object':  # If the feature is categorical
            le_feature = LabelEncoder()
            df_impute_encoded[feature] = le_feature.fit_transform(df_impute[feature].astype(str))

    # Train model only on non-missing data
    model = RandomForestClassifier(n_estimators=100, random_state=42)
    model.fit(df_impute_encoded.loc[~missing_mask], df_impute.loc[~missing_mask, col])

    # Predict missing values
    df_impute.loc[missing_mask, col] = model.predict(df_impute_encoded.loc[missing_mask])

    # Convert back to original labels after prediction using inverse_transform
    df[col] = le.inverse_transform(df_impute[col].astype(int))

# Check missing values again
print("Remaining NaNs after predictive imputation:\n", df[missing_categoricals].isnull().sum())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency

# Step 1: Identify Categorical Columns (Including Target)
categorical_cols = df_clean.select_dtypes(exclude=['number']).columns.tolist()

# Function to compute Cramér's V
def cramers_v(x, y):
    """Calculate Cramér's V for categorical variables."""
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1)) / (n-1))
    r_corr = r - ((r-1)**2) / (n-1)
    k_corr = k - ((k-1)**2) / (n-1)
    return np.sqrt(phi2corr / min(r_corr-1, k_corr-1))

# Compute Cramér's V Matrix for All Categorical Variables
cramers_v_matrix = pd.DataFrame(index=categorical_cols, columns=categorical_cols)

for col1 in categorical_cols:
    for col2 in categorical_cols:
        if col1 == col2:
            cramers_v_matrix.loc[col1, col2] = 1.0  # Perfect correlation with itself
        else:
            cramers_v_matrix.loc[col1, col2] = cramers_v(df_clean[col1], df_clean[col2])

cramers_v_matrix = cramers_v_matrix.astype(float)

# Display Categorical Correlation Matrix
print("\nCramér's V Correlation Matrix for Categorical Variables:")
print(cramers_v_matrix)

# Step 2: Plot Heatmap for Full Categorical Correlation
plt.figure(figsize=(12, 8))
sns.heatmap(cramers_v_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Cramér's V Heatmap: Categorical Variables vs Each Other")
plt.show()
```

Logistic Regression


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Define features and target
X = final_df_reduced_df.drop(columns=['sfdm2_numeric'])  # Drop target column
y = final_df_reduced_df['sfdm2_numeric']  # Target variable (should have more than 2 classes)

# Split data into train-test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Initialize and train Multi-Class Logistic Regression model
log_reg = LogisticRegression(multi_class='multinomial', class_weight='balanced', solver='lbfgs', max_iter=500)
log_reg.fit(X_train_scaled, y_train)

# Predict and evaluate
y_pred = log_reg.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)

# Print accuracy and classification report
print(f'Accuracy: {accuracy:.4f}')
print(classification_report(y_test, y_pred))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)

# Plot Confusion Matrix using Seaborn heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()




```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from itertools import cycle

# Binarize the labels (One-vs-Rest)
y_test_binarized = label_binarize(y_test, classes=np.unique(y_test))
n_classes = y_test_binarized.shape[1]  # Number of classes

# Get probability scores
y_score = log_reg.predict_proba(X_test_scaled)

# Initialize FPR, TPR, and AUC dictionaries
fpr, tpr, roc_auc = {}, {}, {}

# Create a plot for the ROC curves
plt.figure(figsize=(10, 7))
colors = cycle(['blue', 'red', 'green', 'purple', 'orange'])  # Define colors for each class

# Compute ROC for each class
for i, color in zip(range(n_classes), colors):
    fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_score[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'Class {np.unique(y_test)[i]} (AUC = {roc_auc[i]:.2f})')

# Compute Macro AUC (average of all AUCs)
macro_auc = np.mean(list(roc_auc.values()))

# Compute Weighted AUC (weighted by class frequency)
class_counts = np.array([(y_test == cls).sum() for cls in np.unique(y_test)])
weighted_auc = np.sum([roc_auc[i] * class_counts[i] / len(y_test) for i in range(n_classes)])

# Plot ROC diagonal (random classifier)
plt.plot([0, 1], [0, 1], 'k--', lw=2, label='Random')

# Labels and Title
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Multi-Class ROC Curve')
plt.legend(loc='lower right')
plt.show()

# Print AUC scores
print(f"Macro AUC Score: {macro_auc:.4f}")
print(f"Weighted AUC Score: {weighted_auc:.4f}")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import cohen_kappa_score

# Define features and target
X = final_df_reduced_df.drop(columns=['sfdm2_numeric'])  # Drop target column
y = final_df_reduced_df['sfdm2_numeric']  # Target variable (should have more than 2 classes)

# Initialize KFold
n_splits = 5  # Number of folds
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

# Initialize lists to store metrics
accuracies = []
f1_scores = []
classification_reports = []
confusion_matrices = []
qwk_scores = []
custom_scores = []

# Initialize Logistic Regression model
log_reg = LogisticRegression(multi_class='multinomial', class_weight='balanced', solver='lbfgs', max_iter=500)

# K-Fold Cross-validation
for fold, (train_index, test_index) in enumerate(kf.split(X), start=1):
    print(f"Training fold {fold}/{n_splits}...")

    # Split data into training and test sets for this fold
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Standardize features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Train the Logistic Regression model
    log_reg.fit(X_train_scaled, y_train)

    # Predict and evaluate
    y_pred = log_reg.predict(X_test_scaled)

    # Calculate accuracy for this fold
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)

    # Classification report
    report = classification_report(y_test, y_pred, output_dict=True)
    classification_reports.append(report)

    # F1 score (macro average)
    f1 = report['macro avg']['f1-score']
    f1_scores.append(f1)

    # Confusion matrix
    conf_matrix = confusion_matrix(y_test, y_pred)
    confusion_matrices.append(conf_matrix)

    # Quadratic Weighted Kappa (QWK)
    qwk = cohen_kappa_score(y_test, y_pred, weights='quadratic')
    qwk_scores.append(qwk)

    # Custom Score (Average of F1 and QWK)
    custom_score = (f1 + qwk) / 2
    custom_scores.append(custom_score)

    # Plot confusion matrix for this fold
    plt.figure(figsize=(8, 6))
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
    plt.title(f'Confusion Matrix - Fold {fold}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

# Print Average Metrics across all folds
print(f"\nAverage Accuracy across {n_splits} folds: {np.mean(accuracies):.4f}")
print(f"Average F1 Score (Macro Average) across {n_splits} folds: {np.mean(f1_scores):.4f}")
print(f"Average Quadratic Weighted Kappa (QWK) across {n_splits} folds: {np.mean(qwk_scores):.4f}")
print(f"Average Custom Score (F1 + QWK)/2 across {n_splits} folds: {np.mean(custom_scores):.4f}")

# Optionally, print average classification report (macro and weighted averages)
print("\nAverage Classification Report (Macro Average):")
avg_report = np.mean([np.array(list(report['macro avg'].values())) for report in classification_reports], axis=0)
print(f"Precision: {avg_report[0]:.4f}, Recall: {avg_report[1]:.4f}, F1-score: {avg_report[2]:.4f}")
```
